# Explore: data_year_team_franchise.csv

```python
import pandas as pd

df = pd.read_csv("data/data_year_team_franchise.csv")
print(f"Shape: {df.shape}")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
# Check target variable and missing values
print("Target (W) stats:")
print(df["W"].describe())
print(f"\nMissing values:\n{df.isnull().sum()[df.isnull().sum() > 0]}")
print(f"\nUnique teams: {df['teamID'].nunique()}")
print(f"Unique franchises: {df['franchID'].nunique()}")
print(f"Year range: {df['yearID'].min()} - {df['yearID'].max()}")
```

```python
# Distribution of win_bins (target categories)
print("win_bins distribution:")
print(df["win_bins"].value_counts().sort_index())
```

```python
# Column categories
bool_cols = [c for c in df.columns if df[c].dtype == 'object' and df[c].isin(['True','False']).any()] or \
            [c for c in df.columns if set(df[c].dropna().unique()) <= {True, False}]
numeric_cols = df.select_dtypes(include='number').columns.tolist()
cat_cols = [c for c in df.columns if c not in numeric_cols]

print(f"Numeric columns ({len(numeric_cols)}): {numeric_cols}")
print(f"\nCategorical columns ({len(cat_cols)}): {cat_cols}")
```

```python
# Correlation with target W
corr_with_W = df[numeric_cols].corr()["W"].drop("W").sort_values(ascending=False)
print("Correlation with W (wins):")
print(corr_with_W.to_string())
```
